```python
# 2.  Write a Python program to find the length of a tuple.

# Input
t = (1, 2, 3, 4, 5)

# Expected Output
#  Length = 5

print(f'Length = {len(t)}')
```

```python
# 3.  Write a Python program to count the occurrences of an element in a tuple.

# Input
tup = (1, 2, 2, 3, 2, 4)
element = 2

# Expected Output
#  Occurrences of 2 = 3

count = 0

for i in tup:
    if i == element:
        count += 1

print(f'Occurences of {element} = {count}')
```

```python
# 4.  Write a Python program to find the index of an element in a tuple.

# Input
t = (10, 20, 30, 40)
element = 30

# Expected Output
#  Index = 2

for i in range(len(t)):
    if t[i] == element:
        print(f'Index = {i}')
```

```python
# 5.  Write a Python program to convert a tuple into a list and a list into a tuple.

# Input
t = (1, 2, 3)
li = [4, 5, 6]

# Expected Output
#  Tuple to List = [1, 2, 3]
# List to Tuple = (4, 5, 6)

print(f'Tuple to List = {list(t)}')
print(f'List to Tuple  = {tuple(li)}')
```

```python
# 6.  Write a Python program to find the maximum and minimum values in a tuple.

# Input
t = (8, 3, 15, 1, 9)

# Expected Output
#  Maximum = 15, Minimum = 1

print(f'Maximum = {max(t)}, Minimum = {min(t)}')
```
