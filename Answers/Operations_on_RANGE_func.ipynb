```python
# 1.  Write a Python program to print Pascal's triangle for n rows using nested range() loops.

# Input
rows = 5

# Expected Output
# 1
# 1 1
# 1 2 1
# 1 3 3 1
# 1 4 6 4 1
result = [[1]]

for i in range(rows - 1):
    for j in range(i+1, rows):
        result.append([i+j])

print(result)
```
