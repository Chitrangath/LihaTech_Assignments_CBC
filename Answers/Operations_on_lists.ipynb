```python
# 4.  Write a Python program to count even and odd numbers in a list.

# Input
li = [1, 2, 3, 4, 5, 6]

# Expected Output
#  Even count = 3, Odd count = 3

even_cnt = 0
odd_cnt = 0

for i in li:
    if i % 2 == 0:
        even_cnt += 1
    if i %2 != 0:
        odd_cnt += 1

print(f"Even count: {even_cnt}, Odd Count: {odd_cnt}")
```

```python
# 6.  Write a Python program to reverse a list without using the reverse() method.

# Input
li = [10, 20, 30, 40]
n = len(li)
# Expected Output
#  [40, 30, 20, 10]

reversed_li = []
for i in range (n-1, -1, -1):
    if li[i] not in reversed_li:
        reversed_li.append(li[i])
        
print(reversed_li)
```
