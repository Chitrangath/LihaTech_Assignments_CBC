```python
# 14) Python program to find the first non-repeating character in a string

string = "aabbcddeee"

frequency_of_chars = {}

for ch in string:
    frequency_of_chars[ch] = frequency_of_chars.get(ch,0) + 1

for ch in string:
    if frequency_of_chars[ch] == 1:
        print(f"First non repeating character:  {ch}")
        break
```

```python
# 15) Python program to remove duplicate characters from a string while preserving order.

string = "programming"

seen = set()

result = ""

for ch in string:
    if ch not in seen:
        seen.add(ch)
        result = result + ch

print(result)

```

```python
# 16 ) Python program to count uppercase letters, lowercase letters, digits, and special characters in a string.

string = "LihaTech@2026"

uppercase_lettters = ""
lowercase_letters = ""
digits = ""
special_char = ""


for ch in string:
    if ch.isupper():
        uppercase_lettters += ch
    elif ch.islower():
        lowercase_letters += ch
    elif ch.isdigit():
        digits += ch
    else:
        special_char += ch

print("uppercase_lettters: ",uppercase_lettters)
print("lowercase_letters: ",lowercase_letters)
print("Digits: ",digits)
print("special_char: ",special_char)
```
